```python
monthly_adr = df.groupby('arrival_date_month')['adr'].mean()
```

```python
plt.figure(figsize=(13,5))
plt.grid(alpha=0.5)
plt.gca().set_facecolor('pink')
plt.plot(monthly_adr.index,
         monthly_adr.values,
         marker='o',
         color='blue',
         markerfacecolor='r',
         markeredgecolor='blue',
         label='line plot')
plt.legend()
plt.title("Average ADR by Month")
plt.xlabel("Arrival Month")
plt.ylabel("Average Daily Rate (ADR)")
plt.show()
```

```python
plt.figure(figsize=(6,6),facecolor='white')

plt.gca().set_facecolor('skyblue')
a=sns.countplot(x='hotel', data=df,color='pink',edgecolor='black')

plt.title("Number of Bookings by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Number of Bookings")
for i in a.containers: 
   a.bar_label(i,label_type='edge',color='red')   

plt.show()
```

```python
hotel_count =df['hotel'].value_counts()
```

```python
plt.figure(figsize=(6,6),facecolor='pink')
plt.pie(
    hotel_count.values,
    labels=hotel_count.index,
    autopct='%1.1f%%',
    startangle=90,
    explode=(0.03, 0),shadow=True,
    colors=['grey','orange' ],
    textprops={'color':'white'}
)
plt.title("Percentage of Bookings by Hotel Type")

plt.show()
```

```python
plt.figure(figsize=(8,5),facecolor='white')
plt.gca().set_facecolor('pink')
plt.grid(alpha=0.5)
plt.hist(
    df['lead_time'],
    bins=10,
    edgecolor='black',
    color='grey',
    rwidth=0.96,
)
plt.title('Distribution of Lead Time')
plt.xlabel('Lead Time')
plt.ylabel('Number of Bookings')
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.grid(alpha=0.5)
sns.barplot(
    x='hotel',
    y='adr',
    data=df,errorbar=None
)
plt.title("Average ADR by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")
for i in a.containers:
    a.bar_label(i,label_type='edge',color='red',rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    x='hotel',
    y='adr',
    data=df
)
plt.title("Distribution of ADR by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")

plt.show()
```

```python
q1 = df['adr'].quantile(0.25)
q3 = df['adr'].quantile(0.75)
```

```python
q1
```

```python
q3
```

```python
iqr = q3 - q1
```

```python
lower= q1 - 1.5 * iqr
upper= q3 + 1.5 * iqr
```

```python
lower
```

```python
upper
```

```python
outliers=df[(df['adr']>upper) | (df['adr']<lower)] 
outliers.info()
```

```python

non_outliers = df[(df['adr'] >lower) & (df['adr'] <upper)]
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    x='hotel',
    y='adr',
    data=non_outliers
)
plt.ylim(0,800)
```

```python
print(df['adr'].max())
print(non_outliers['adr'].max())
```

```python
plt.figure(figsize=(7,5))
sns.boxplot(
    x='hotel',
    y='adr',
    data=non_outliers,
    showfliers=False
)
plt.title("Distribution of ADR by Hotel Type")
plt.xlabel("Hotel Type")
plt.ylabel("Average Daily Rate")

plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x='lead_time',
    y='adr',
    data=df,
)
plt.ylim(0,600)
plt.title("Lead Time vs Average Daily Rate")
plt.xlabel("Lead Time (Days)")
plt.ylabel("Average Daily Rate")

plt.show()
```
